# TCGAbiolinks

```r
library(TCGAbiolinks)
```

```r
TCGA_PROJECT <- "TCGA-COAD"
DATA_DIR <- "~/dataset_1/database/GDCdata"
```

```r
# Get Project Summary from GDC
getProjectSummary(TCGA_PROJECT)
```

```r
# Download clinical information
clinical <- GDCquery_clinic(project = TCGA_PROJECT, type = "clinical")

dim(clinical)

clinical[1:6, 1:6]
```

```r
# Gene Expression Quantification
query <- GDCquery(
    project = TCGA_PROJECT,
    data.category = "Transcriptome Profiling",
    data.type = "Gene Expression Quantification",
    workflow.type = "STAR - Counts")
```

```r
GDCdownload(query, directory = DATA_DIR)
```

```r
eSet <- GDCprepare(query, directory = DATA_DIR)
```

```r
eSet
```

```r
# DNA Methylation Array
query <- GDCquery(
    project = TCGA_PROJECT,
    data.category = "DNA Methylation",
    data.type = "Methylation Beta Value",
    platform = "Illumina Human Methylation 450"
)
```

```r
# The API may be interrupted when downloading large amounts of data.
GDCdownload(query, directory = DATA_DIR)
```
